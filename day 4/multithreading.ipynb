```python
import time
import threading

def op_1(d, q):
    time.sleep(7)
    print(f"op 1 {d} {d}")
def op_2():
    time.sleep(5)
    print("op 2 done")
def op_3():
    time.sleep(2)
    print("op 3 done")


th_1 = threading.Thread(target=op_1, args=('d', 'b'))
th_2 = threading.Thread(target=op_2)
th_3 = threading.Thread(target=op_3)

th_1.start() # for running the thread
th_2.start()
th_3.start()

th_1.join()
print('done') # runs only after the third thread
th_2.join()
th_3.join()
```
