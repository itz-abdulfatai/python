```python
import time

print('the time is ', time.time())
```

```python
def timer_dec(base_func):
    def enhanced_fn():
        start_time = time.time()
        base_func()
        end_time = time.time()
        print(f"it took {end_time - start_time}s")
    return enhanced_fn



@timer_dec
def brew_tea():
    print("brewing tea")
    time.sleep(1)
    print("tea is ready")
# timed_brew_tea = timer_dec(brew_tea)
# timed_brew_tea()
brew_tea()
```
