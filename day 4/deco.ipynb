```python
import time


def mission_timer(func):
    def wrapper(*args, **kwargs):
        start_time = time.time()
        print(3)
        print(2)
        print(1)
        return_value = func(*args, **kwargs)
        end_time=time.time()
        print(f"mission duration: {end_time-start_time} secs")
        return return_value


    return wrapper

@mission_timer
def launch_probe(target):
    print(f"Launching probe toward {target}...")
    time.sleep(1)
    return f"Probe successfully en route to {target}."

@mission_timer
def deploy_satellite(orbit_type):
    print(f"Deploying satellite into {orbit_type} orbit")
    time.sleep(2)

result = launch_probe("Europa")
print(f"launch result is {result}")
```
