```python
import time
from multiprocessing import Process

def t_1():
    time.sleep(5)
    print("t1 done")
def t_2():
    time.sleep(1)
    print("t2 done")
def t_3():
    time.sleep(2)
    print("t3 done")

m_1 = Process(target=t_1)
m_2 = Process(target=t_2)
m_3 = Process(target=t_3)



m_1.start()
m_2.start()
m_3.start()
m_1.join()
m_2.join()
m_3.join()
print("done")
```
