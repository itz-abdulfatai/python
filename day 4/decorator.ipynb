```python
import time
# problem 1 (https://colab.research.google.com/drive/1ZYPNRo5rtPYIArHP9AvLWimxH93uGr4f?usp=sharing#scrollTo=0nTbeNwUA94L)
def mission_timer(base_func):
    def aug_func():
        start_time = time.time()
        print(3)
        print(2)
        print(1)
        base_func()
        end_time=time.time()
        print(f"mission duration: {end_time-start_time} secs")

    return aug_func

@mission_timer 
def launch_probe():
    print("launching probe")
    time.sleep(1)

@mission_timer 
def deploy_satellite():
    print("deploying satellite")
    time.sleep(2)

launch_probe()
deploy_satellite()
```

```python
# problem 2
def mission_timer(base_func):
    def aug_func(*args, **kwargs):
        start_time = time.time()
        print(3)
        print(2)
        print(1)
        base_func(*args, **kwargs)
        end_time=time.time()
        print(f"mission duration: {end_time-start_time} secs")

    return aug_func

@mission_timer
def launch_probe(target):
    print(f"Launching probe toward {target}...")
    time.sleep(1)

@mission_timer
def deploy_satellite(orbit_type, altitude_km):
    print(f"Deploying satellite into {orbit_type} orbit at {altitude_km} km...")
    time.sleep(2)

launch_probe("Mars")
deploy_satellite("polar", 800)
deploy_satellite(orbit_type="geostationary", altitude_km=35786)

```

```python
# problem 3
def mission_timer(base_func):
    def aug_func(*args, **kwargs):
        start_time = time.time()
        print(3)
        print(2)
        print(1)
        return_value = base_func(*args, **kwargs)
        end_time=time.time()
        print(f"mission duration: {end_time-start_time} secs")
        return return_value

    return aug_func

@mission_timer
def launch_probe(target):
    print(f"Launching probe toward {target}...")
    time.sleep(1)
    return f"Probe successfully en route to {target}."

@mission_timer
def deploy_satellite(orbit_type, altitude_km):
    print(f"Deploying satellite into {orbit_type} orbit at {altitude_km} km...")
    time.sleep(2)

deploy_satellite("polar", 800)

result = launch_probe("Europa")
print("Stored mission result:", result)
```
